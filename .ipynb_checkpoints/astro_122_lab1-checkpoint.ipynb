```python
import numpy as np
import matplotlib.pyplot as plt
from joce_radiolab.sampling import power_spectrum

fs = 2_000_000
Ns = [2048, 4096, 8192, 16384]

plt.figure(figsize=(9,5))

for N in Ns:
    fname = f"FreqRes{fs}_Hz, {N}.npz"
    z = np.load(fname)
    x = z["data"][:N].astype(float)

    freqs, P = power_spectrum(x, fs, shift=True)
    P /= np.max(P)   # normalize for shape comparison

    df = fs / N
    plt.plot(freqs/1e6, P, label=f"N={N}, Δf={df:.1f} Hz")

plt.xlabel("Frequency (MHz)")
plt.ylabel("Normalized power")
plt.title("Effect of dense frequency spacing on spectral leakage")
plt.xlim(-0.6, 0.6)  # zoom around the tone
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import joce_radiolab.sampling as s

code = """import numpy as np

def load_npz(filename):
    \\"\\"
    Load SDR data saved as .npz.
    Expected keys: 'data', 'sample_rate'
    \\"\\"
    z = np.load(filename)
    return z["data"], float(z["sample_rate"])

def power_spectrum(x, fs, shift=True):
    \\"\\"
    Compute power spectrum |FFT|^2.
    Returns (freqs, power).
    \\"\\"
    x = np.asarray(x)
    X = np.fft.fft(x)
    freqs = np.fft.fftfreq(len(x), d=1/fs)
    P = np.abs(X)**2

    if shift:
        freqs = np.fft.fftshift(freqs)
        P = np.fft.fftshift(P)

    return freqs, P
"""

open(s.__file__, "w").write(code)
print("WROTE:", s.__file__)
```
