```python
import numpy as np
import matplotlib.pyplot as plt
from joce_radiolab.sampling import power_spectrum

fs = 2_000_000
Ns = [2048, 4096, 8192, 16384]

plt.figure(figsize=(9,5))

for N in Ns:
    fname = f"FreqRes{fs}_Hz, {N}.npz"
    z = np.load(fname)
    x = z["data"][:N].astype(float)

    freqs, P = power_spectrum(x, fs, shift=True)
    P /= np.max(P)   # normalize for shape comparison

    df = fs / N
    plt.plot(freqs/1e6, P, label=f"N={N}, Δf={df:.1f} Hz")

plt.xlabel("Frequency (MHz)")
plt.ylabel("Normalized power")
plt.title("Effect of dense frequency spacing on spectral leakage")
plt.xlim(-0.6, 0.6)  # zoom around the tone
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
